# Глава 11. Специализированные применения

## 11.1. RAG в юридической сфере

RAG-технологии меняют юридическую практику: они ускоряют поиск правовой информации и помогают принимать решения. Раньше юристы часами изучали документы, теперь системы находят нужные нормы и судебные решения за секунды.

Правовые системы сложны, поэтому RAG здесь особенно полезен. Законодательство включает федеральные законы, подзаконные акты, региональные нормы и судебную практику. Юристу нужно учитывать всё это одновременно, а также отслеживать изменения. Человеческая память с такой нагрузкой справляется всё хуже.

Юридические вопросы многогранны. Один случай может рассматриваться с точки зрения разных отраслей права, требовать анализа множества источников и учёта конкретных обстоятельств. Обычные RAG-системы часто неточны в юриспруденции, потому что не учитывают эти нюансы.

Специализированные юридические RAG-системы используют гибридный поиск: одновременно ищут в законах, судебных решениях, доктринах и разъяснениях, присваивая каждому источнику вес в зависимости от типа вопроса. Это даёт комплексный анализ, учитывающий и нормы, и практику.

MultiQueryRetriever повышает эффективность: система генерирует несколько формулировок одного запроса с разными юридическими терминами и описаниями обстоятельств. Это улучшает полноту поиска.

Результаты впечатляют. Анализ прецедентов, который раньше занимал часы, теперь выполняется за минуты. Качество часто выше ручного поиска: система обрабатывает больше документов и не устаёт.

Бенчмарки вроде LexRAG задают стандарты качества для юридических ИИ. Они включают тысячи диалогов, размеченных юристами, и проверяют точность правовой квалификации и применимость советов. Российские разработчики создают аналогичные бенчмарки для отечественного права.

Корпоративные юристы применяют RAG для рутины. Банки оценивают правовые риски сделок, страховые — согласуют медицинские услуги, производители — проверяют контракты на соответствие стандартам. Обработка типовых запросов сократилась с часов до минут.

Судебная практика как источник данных сложна: она меняется и противоречива. Высшие суды могут изменить толкование норм, поэтому системы должны обновлять базы и переоценивать старые решения. Современные RAG-системы отслеживают изменения практики и корректируют веса документов.

Этика — спорный вопрос. Американская ассоциация адвокатов напоминает: использование ИИ не снимает с юриста ответственности. Были случаи, когда ИИ ссылался на несуществующие решения, поэтому контроль человека обязателен.

Юридическая терминология требует особой векторизации. Правовые тексты полны специальных терминов, латыни и отсылок. Обычные модели эмбеддингов плохо понимают юридический язык, поэтому нужны специализированные, например LegalBERT.

GraphRAG эффективен в праве, потому что связи между нормами важны. Граф знаний моделирует иерархию законов, подзаконных актов и решений, а также связи между правовыми концепциями. Это позволяет находить не только прямые, но и косвенные связи, важные для квалификации.

Интеграция RAG с OCR помогает работать с архивными документами. Многие решения существуют только на бумаге. Системы оцифровывают их, индексируют и делают доступными для поиска, расширяя базу источников.

Будущее — за более специализированными и точными системами. Они смогут не только находить нормы, но и проводить юридический анализ, выявлять коллизии и предлагать стратегии защиты. Для этого нужно сотрудничество ИТ-специалистов и юристов.
