# Глава 9. Память и контекст

## 9.1. Подсистемы памяти для ИИ‑агентов

Подсистема памяти — ключевой компонент ИИ-агентов, обеспечивающий обучение, адаптацию и персонализацию. В отличие от статичных моделей, такие агенты накапливают опыт и формируют долгосрочные отношения с пользователем.

Объединение памяти с RAG даёт синергию: внешние знания дополняют личный опыт агента. Это позволяет строить адаптивные системы, использующие и базы знаний, и историю взаимодействий.

Архитектура памяти делится на краткосрочную и долгосрочную, имитируя когнитивные процессы человека при принятии решений и генерации ответов.

### Краткосрочная память и рабочий контекст

**Краткосрочная память** — это рабочее пространство агента для текущей задачи. Включает три элемента:

- **Рабочая память** — временное хранение данных и промежуточных результатов. В отличие от статичного контекстного окна, динамически структурирует информацию, выделяет приоритеты и связывает части контекста.
- **Буфер внимания** — управляет фокусом: определяет приоритетные аспекты задачи, активирует нужные воспоминания и распределяет вычислительные ресурсы.
- **Сессионная память** — поддерживает непрерывность диалога: отслеживает цели пользователя, промежуточные результаты, контекст и эмоциональную динамику.

### Эпизодическая память: хранение опыта

**Эпизодическая память** фиксирует события и взаимодействия в хронологическом порядке, позволяя агенту учиться на прошлом опыте. Включает:

- **Историю диалогов** — структурированное хранение разговоров с анализом интенций, эмоций, успешности и паттернов поведения пользователя, а не просто архивирование текста.
- **Траектории действий** — документирование решений и их результатов. Агент анализирует успешные стратегии и применяет этот опыт в будущем.
- **Временную индексацию** — быстрый поиск событий по хронологии, анализ изменений во времени и выявление сезонных паттернов поведения.

### Семантическая память: структурированные знания

**Семантическая память** — это сеть фактов и концепций, которая, в отличие от параметрических знаний модели, динамически обновляется на основе нового опыта. Состоит из:

- **Базы фактов** — проверенная информация о мире и пользователях. Факты аннотируются источником, достоверностью, временными рамками и связями.
- **Концептуальных связей** — семантическая сеть для понимания отношений между понятиями, построения выводов, аналогий и переноса знаний между областями.
- **Онтологий** — формальные структуры знаний в предметных областях: классы объектов, их свойства и отношения. Обеспечивают согласованность и логичность рассуждений агента.

### Процедурная память: навыки и алгоритмы

**Процедурная память** хранит знания о выполнении задач: алгоритмы, навыки работы с инструментами и успешные поведенческие паттерны. Включает:

- **Навыки и умения** — исполняемые процедуры. Агент обучается им через наблюдение и совершенствует через практику и обратную связь.
- **Алгоритмы решений** — эффективные стратегии для типовых задач. Система адаптирует их под конкретные условия и комбинирует для сложных проблем.
- **Паттерны поведения** — успешные стратегии взаимодействия с разными пользователями и ситуациями. Агент распознаёт контекст и автоматически применяет подходящий паттерн.

### Эмоциональная память: персонализация и социальный контекст

**Эмоциональная память** персонализирует взаимодействие через понимание предпочтений, настроений и социального контекста. Критична для долгосрочных отношений с пользователем. Включает:

- **Предпочтения пользователя** — анализ явных и неявных сигналов для адаптации стиля общения, контента и стратегий.
- **Эмоциональный контекст** — оценка настроения и состояния пользователя, адаптация тона и проявление эмпатии.
- **Социальные связи** — моделирование отношений, иерархии и групповой динамики для адаптации поведения к социальному контексту.

### Ассоциативно‑гетерархическая память: объединение всех видов памяти

**Ассоциативно-гетерархическая память** — интегрированная архитектура, объединяющая все типы памяти в единую систему. В отличие от жёсткой иерархии, допускает динамические связи без фиксированного подчинения уровней.

**Ключевые принципы:**

- **Ассоциативность** — автоматические связи между семантически близкими элементами разных типов памяти (эпизоды, факты, навыки, предпочтения) на основе контекстуального сходства.
- **Гетерархия** — любой тип памяти может становиться ведущим в зависимости от контекста. Эмоции способны переопределить факты, процедуры — эпизоды.
- **Интеграция** — кросс-модальные связи через векторные представления и языковые модели. Новые воспоминания автоматически связываются с релевантными существующими.
- **Динамическая активация** — контекстуально-зависимое извлечение с одновременной активацией разных типов памяти и каскадным распространением по связям.
- **Временная консолидация** — перевод частых комбинаций воспоминаний в устойчивые связи. Забывание координируется между всеми типами памяти.
- **Адаптивная реструктуризация** — укрепление сильных связей, ослабление слабых, формирование новых на основе повторяющихся паттернов.

**Интеграция с RAG** — внутренняя ассоциативная память дополняет векторный поиск внешних знаний, выявляя неочевидные связи между личным опытом агента и глобальной информацией.

**Результат** — эмерджентные свойства системы превосходят сумму возможностей отдельных типов памяти: гибкость, творческие ассоциации, адаптивное поведение. Это основа для сложного когнитивного поведения, персонализации и непрерывного обучения.

### Интеграция с RAG‑системами

**RAG как внешнее расширение памяти** — векторный поиск дополняет внутренние механизмы, позволяя агенту извлекать информацию и из личного опыта, и из внешних источников.

**Семантический поиск** — объединённый анализ: факты из семантической памяти, эпизоды из истории взаимодействий и внешние данные через RAG запрашиваются одновременно.

**Динамическое управление контекстом** — менеджер памяти определяет оптимальное сочетание внутренних воспоминаний и внешних знаний для формирования контекста языковой модели.

**Итог:** интеграция всех компонентов памяти с RAG создаёт основу для интеллектуальных, адаптивных агентов, способных к обучению, персонализации и долгосрочным отношениям с пользователями.

## 9.2. Управление долгосрочным и краткосрочным контекстами

Управление контекстом в ИИ-агентах — сложная задача. Нужно совместить ограничения языковых моделей и потребность в долгой памяти. Система должна легко переключаться между мгновенной реакцией и накоплением знаний.

Разница между короткой и долгой памятью показывает главные ограничения современных моделей. Контекстное окно даёт быстрый доступ к данным, но оно маленькое. Для долгой памяти нужны особые способы хранения, поиска и сжатия информации.

### Архитектура управления контекстом

**Как работает память ИИ-агента (три уровня)**

У контекста есть три уровня:
1. **Рабочая память** — то, что используется прямо сейчас.
2. **Буфер** — то, что было важно недавно.
3. **Долгая память** — старые данные и накопленные знания.

**Кто всем управляет**

Менеджер контекста решает, какую информацию держать под рукой в данный момент. Он оценивает три вещи: насколько данные полезны, как давно их использовали и насколько они важны для текущей задачи.

**Как система выбирает важное**

Чтобы понять, что важно, система смотрит на:
- как давно использовали информацию (недавность);
- как часто к ней обращались;
- подходит ли она по смыслу к текущему запросу;
- прямые указания пользователя.
При смене задачи приоритеты могут быстро пересматриваться.

**Как сжать данные, чтобы влезло больше**

Используются простые методы:
- краткий пересказ диалогов;
- вытаскивание только фактов;
- объединение повторяющегося;
- представление информации в виде структуры (главное — детали).

### Краткосрочное управление контекстом

**Как работает рабочая память агента**

Рабочая память — это активное пространство для текущей задачи. В ней лежит то, что важно прямо сейчас: запрос пользователя, промежуточные результаты, запущенные инструменты и их состояние.

**Внимание**

Механизмы внимания управляют фокусом внутри краткосрочной памяти. Система переключает внимание между частями задачи и выделяет самое важное.

**Сохранение промежуточных итогов**

Чтобы не терять прогресс в многошаговых задачах, агент сохраняет частичные решения, выводы и связи между данными. Это пригодится на следующих шагах.

**Удержание контекста внутри сессии**

Система следит за непрерывностью диалога: помнит цели пользователя, эмоциональный фон, связи между фразами и развитие темы.

*(Термин «рецентность» означает оценку по двум признакам: как часто и как давно использовался факт из памяти агента.)*

### Долгосрочное управление контекстом

**Как информация попадает в долгую память**

Консолидация переносит важные данные из короткой памяти в долгую. Система оценивает: как часто это используют, насколько это важно для пользователя, есть ли связь с уже известным и пригодится ли в будущем.

**Как устроен поиск**

Чтобы быстро находить информацию в большой памяти, система строит многоуровневые указатели: по темам, времени, пользователю, эмоциям. Смысловой поиск (через векторные представления) находит близкие по значению данные, даже если слова разные.

**Как забывать лишнее**

Система не копит всё подряд. Она:
- удаляет устаревшее;
- объединяет похожие воспоминания;
- упрощает старые события (меньше деталей);
- архивирует то, чем редко пользуются.

**Как память приводит в порядок**

Время от времени система перестраивает долгую память: смотрит, к чему чаще обращаются, обновляет указатели, связывает родственные воспоминания и пересматривает, что теперь важно.

### Интеграция с RAG‑системами

**Как внешние знания связаны с внутренней памятью (RAG)**

RAG — это мост между своей памятью и внешними источниками. Система сама решает, взять ответ из внутренней памяти или пойти за данными вовне.

**Смешанный поиск**

Запрос уходит сразу в два места: во внутреннюю память и во внешние источники через RAG. Потом результаты сводятся вместе, чтобы получить самый точный контекст.

**Кеширование внешних данных**

Если к каким-то внешним данным обращаются часто, система сохраняет их во внутренней памяти. Так в следующий раз не придётся снова лезть наружу — ответ уже под рукой.

**Что делать с противоречиями**

Данные изнутри и снаружи могут не совпадать. Тогда система сравнивает их и выбирает, чему верить: своему опыту или внешним знаниям. Выбор зависит от контекста и настроек.

### Адаптивные стратегии управления

**Как система подстраивается под задачу**

Стратегии меняются в зависимости от того, что нужно сделать:
- Для творческих задач — берётся больше разнородной информации.
- Для аналитики — наоборот, только самое точное и близкое к делу.

**Учёт привычек пользователя**

Система запоминает предпочтения: как человек любит общаться, насколько подробно объяснять, как часто нужно заглядывать в историю. Под каждого пользователя настройки свои.

**Как меняется важность со временем**

Свежие события ценятся выше. Старые факты теряют вес, особенно если свежие данные их не подтверждают.

**Как агент оценивает сам себя**

Система следит за тем, насколько успешно она управляет памятью. Если какой-то подход работает плохо, она это замечает и перестраивает алгоритмы, чтобы в будущем работа шла лучше.

### Технические реализации и оптимизации

**Как устроена работа памяти без задержек**

Разные уровни памяти обрабатываются по-разному:
- Короткая память — синхронно, чтобы ответ пришёл сразу.
- Долгая память — в фоне, чтобы не тормозить основную работу.

**Как расширять систему**

Память можно разнести по разным серверам. Каждый тип памяти живёт там, где ему удобнее и быстрее работать.

**Как система учится на ходу**

Алгоритмы постепенно улучшаются. Источники обучения: отзывы пользователей и анализ того, насколько успешно прошло общение.

**Как следят за качеством**

Система отслеживает, насколько хорошо работают стратегии управления памятью. Что измеряют:
- скорость ответа;
- насколько точно найдена информация;
- доволен ли пользователь;
- сколько ресурсов потрачено.

**Итог**

Управление короткой и долгой памятью — это живая система, где всё время нужно искать баланс. Хорошая архитектура даёт и быструю реакцию, и способность учиться на долгой дистанции. Именно на этом строится по-настоящему адаптивный и персональный ИИ-агент.
