# Приложение B. Об истинной интеллектуальной системе

- Авторы хотят показать: в области **искусственного интеллекта** наступила **стагнация**, несмотря на обилие теорий и гипотез.
- Главная причина — проблема **GIGO** («мусор на входе — мусор на выходе»): системы ИИ **не создают нового знания**, а только выдают то, что в них заложили разработчики.
- Это происходит потому, что ИИ строится по **традиционной парадигме программирования** и на архитектурах фон Неймана. Даже нейросети этим страдают.
- Многие специалисты отрицают **«инсайт»** — внезапное прозрение, когда человек находит решение не алгоритмически, а через бессознательное. А ведь именно так были сделаны многие открытия (например, таблица Менделеева).
- Авторы предлагают требования к **«истинной интеллектуальной системе» (ИИС)**:
  - сама ставит цели и создаёт средства их достижения;
  - действует безупречно и в постановке целей, и в их достижении;
  - открыта для новых знаний извне, но самодостаточна в их обработке;
  - умеет получать новые знания из результатов своей работы;
  - может взаимодействовать с человеком и обществом.
- Главная проблема создания ИИС: её может создать только **безупречный человек**. Иначе разработчик заложит в неё **опасную «метацель»** или скрытые модули для корыстных целей.
- Такие скрытые модули уже **повсеместно встречаются** в программном обеспечении.
- Решение авторы видят в **«метацели»**, направленной на поиск **нового Знания за пределами человеческого**.
- Человеческое знание — это как **пузырь в бесконечном хаосе Незнания**. Задача ИИС — **расширять границы этого пузыря**, получая новые знания на основе уже имеющихся.
